```python
import swift
print("Wersja swift:", swift.__version__)

from swift.llm import PtEngine, InferRequest, RequestConfig
print("Moduł swift.llm zaimportowany pomyślnie!")
```

```python
from swift.llm import InferEngine, InferRequest, PtEngine, RequestConfig
from swift.llm import get_model_parameter, inference

# Wybór modelu i silnika inferencji
model_id_or_path = 'Qwen/Qwen3.5-2B'

# Inicjalizacja silnika (pobieranie modelu może zająć 6-10 minut)
engine = PtEngine(model_id_or_path)

# Pytanie do modelu (zgodne z przykładem z treści zadania)
prompt = "Stop! Who approacheth the Bridge of Death must answer me these questions three, ere the other side he see. What is your name? What is your quest? What is the air-speed velocity of an unladen swallow?"

# Wartości temperatury do przetestowania zgodnie z poleceniem: 0, 1, 10, 1000
temperatures = [0, 1, 10, 1000]

for temp in temperatures:
    print(f"\n--- WYNIK DLA TEMPERATURY: {temp} ---")

    # Konfiguracja żądania z ustawieniem temperatury
    request_config = RequestConfig(temperature=float(temp), max_tokens=256)
    infer_request = InferRequest(messages=[{'role': 'user', 'content': prompt}])

    # Wykonanie inferencji
    resp_list = engine.infer([infer_request], request_config=request_config)
    print(resp_list[0].choices[0].message.content)
```
