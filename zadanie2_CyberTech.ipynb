```python
from datasets import load_dataset

print("Ładowanie zbioru danych...")
dataset = load_dataset("medalpaca/medical_meadow_wikidoc_patient_information")
print("Liczba przykładów przed filtrowaniem:", len(dataset['train']))

def filter_func(example):
    inp = example.get('input', '') or ''
    out = example.get('output', '') or ''
    if len(out) < 20:
        return False
    if len(inp) + len(out) > 1500:
        return False
    return True

filtered_dataset = dataset['train'].filter(filter_func)
print("Liczba przykładów po filtrowaniu:", len(filtered_dataset))
```

```python
import torch
from swift.pipelines.train.sft import SwiftSft, SftArguments

# Czścimy pamięć GPU
torch.cuda.empty_cache()

args = SftArguments(
    model='Qwen/Qwen3.5-2B',
    dataset=['medalpaca/medical_meadow_wikidoc_patient_information'],
    use_hf=True,
    tuner_type='lora',
    task_type='causal_lm',
    output_dir='output_qwen_lora',
    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,
    learning_rate=1e-4,
    max_length=256,         # Krótki kontekst gwarantujący brak błędu OOM
    max_steps=30,           # Krótki test na 30 kroków do szybkiego zakończenia
    save_steps=30,
    logging_steps=5,
    gradient_checkpointing=True # Kluczowe dla oszczędności pamięci bez bitsandbytes
)

print("Uruchamianie treningu LoRA...")
trainer = SwiftSft(args)
trainer.main()
```
